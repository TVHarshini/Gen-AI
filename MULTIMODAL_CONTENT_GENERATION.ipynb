```python
# ============================================================
# EXPERIMENT 11
# MULTIMODAL CONTENT GENERATION
# TEXT + IMAGE + AUDIO
# ============================================================

!pip install -q transformers diffusers accelerate torch scipy

import torch
from transformers import pipeline
from diffusers import StableDiffusionPipeline
from IPython.display import display, Audio


# ============================================================
# 1. TEXT GENERATION
# ============================================================

print("Generating text...")

text_generator = pipeline(
    "text-generation",
    model="gpt2"
)

prompt = "Artificial Intelligence will change the world by"

result = text_generator(
    prompt,
    max_new_tokens=40,
    num_return_sequences=1
)

print("\n==============================")
print("TEXT GENERATION")
print("==============================")
print(result[0]["generated_text"])


# ============================================================
# 2. IMAGE GENERATION
# ============================================================

print("\n==============================")
print("IMAGE GENERATION")
print("==============================")

if torch.cuda.is_available():

    print("GPU detected.")

    image_pipe = StableDiffusionPipeline.from_pretrained(
        "runwayml/stable-diffusion-v1-5",
        torch_dtype=torch.float16
    )

    image_pipe = image_pipe.to("cuda")

    image_prompt = "A futuristic AI laboratory with robots, digital art"

    image = image_pipe(
        image_prompt,
        num_inference_steps=15
    ).images[0]

    display(image)

    image.save("generated_ai_lab.png")

    print("Image saved as generated_ai_lab.png")

else:

    print("GPU is not available.")
    print("Skipping image generation.")
    print("Enable T4 GPU in Runtime → Change runtime type if required.")


# ============================================================
# 3. AUDIO GENERATION
# ============================================================

print("\n==============================")
print("AUDIO GENERATION")
print("==============================")

audio_generator = pipeline(
    "text-to-speech",
    model="facebook/mms-tts-eng"
)

audio_text = "Artificial Intelligence is transforming the future."

audio_result = audio_generator(audio_text)

audio = audio_result["audio"]
sampling_rate = audio_result["sampling_rate"]

display(
    Audio(
        audio,
        rate=sampling_rate
    )
)

print("Audio generated successfully.")


# ============================================================
# END
# ============================================================

print("\n==============================")
print("EXPERIMENT 11 COMPLETED")
print("==============================")
```
